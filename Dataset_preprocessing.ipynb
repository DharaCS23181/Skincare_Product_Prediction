```python
newdf['notable_effects'] = (
    newdf['notable_effects']
    .str.lower()
    .str.replace('\n','')
    .str.strip()
)
```

```python
newdf['notable_effects'] = newdf['notable_effects'].str.split(',')
```

```python
newdf['notable_effects'] = newdf['notable_effects'].apply(
    lambda x: [i.strip() for i in x]
)
```

```python
effects = set()

for row in newdf['notable_effects']:
    effects.update(row)

print(sorted(effects))
```

```python
newdf.head()
```

```python
effect_condition_map = {
'acne-free' : 'acne',
'acne-spot' : 'acne',

'anti-aging' : 'wrinkles',

'balancing' : 'skin imbalance',

'black-spot' : 'pigmentation',

'brightening' : 'pigmentation',

'hydrating' : 'dry and dehydrated skin',
'moisturizing' : 'dry and dehydrated skin',

'oil-control' : 'oily skin',

'pore-care' : 'enlarged pores',

'skin-barrier' : 'impaired skin barrier',

'soothing' : 'redness',

'refreshing' : 'dull skin',

'uv-protection' : 'sun damage',
'no-whitecast' : 'sun protection'
}
```

```python
def map_condition(effects):

    conditions = []

    for e in effects:
        if e in effect_condition_map:
            conditions.append(effect_condition_map[e])

    return list(set(conditions))

newdf['skin_condition'] = newdf['notable_effects'].apply(map_condition)
```
